```python

# ============================================================
# Stage 07 - Interactive Prediction (Try It)
# Step 7.1 - Setup & Configuration
# ============================================================

from pathlib import Path
import json
import joblib
import numpy as np
import pandas as pd
import torch

from sentence_transformers import SentenceTransformer

# Locate project directory
CURRENT_DIR = Path.cwd().resolve()

PROJECT_DIR = next(
    (
        p for p in [CURRENT_DIR, *CURRENT_DIR.parents]
        if (p / "models" / "traditional_tfidf_logreg.joblib").exists()
    ),
    None
)

assert PROJECT_DIR is not None, "Project directory not found!"

# Model paths
MODEL_DIR = PROJECT_DIR / "models"
MINILM_DIR = MODEL_DIR / "minilm"

TRADITIONAL_PATH = (
    MODEL_DIR / "traditional_tfidf_logreg.joblib"
)

ENCODER_PATH = MINILM_DIR / "sentence_transformer"
CLASSIFIER_PATH = MINILM_DIR / "classifier.joblib"
CONFIG_PATH = MINILM_DIR / "config.json"

# Verify required files
assert TRADITIONAL_PATH.exists()
assert ENCODER_PATH.exists()
assert CLASSIFIER_PATH.exists()
assert CONFIG_PATH.exists()

# Device
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("=" * 55)
print("STAGE 07 - INTERACTIVE PREDICTION")
print("=" * 55)

print("Project:", PROJECT_DIR)
print("Device:", DEVICE)
print("Required Files: FOUND")

print("\nSetup Verification: PASSED!")
```

```python

# ============================================================
# Step 7.1 - Load Trained Models
# ============================================================

# 1. Load traditional pipeline
traditional_model = joblib.load(
    TRADITIONAL_PATH
)

print("Traditional Model: LOADED")

# 2. Load MiniLM encoder
minilm_encoder = SentenceTransformer(
    str(ENCODER_PATH),
    device=DEVICE
)

print("MiniLM Encoder: LOADED")

# 3. Load MiniLM classifier
minilm_classifier = joblib.load(
    CLASSIFIER_PATH
)

print("MiniLM Classifier: LOADED")

# 4. Load configuration
with open(CONFIG_PATH, "r", encoding="utf-8") as f:
    minilm_config = json.load(f)

# 5. Verify matching target classes
assert set(traditional_model.classes_) == set(
    minilm_classifier.classes_
)

assert len(traditional_model.classes_) == 5

# 6. Verify MiniLM configuration
assert minilm_config["embedding_dimension"] == 384
assert minilm_config["normalize_embeddings"] is True

print("\n" + "=" * 55)
print("MODEL LOADING VERIFICATION")
print("=" * 55)

print("Traditional Model: READY")
print("MiniLM Model: READY")

print("\nSupported Classes:")
for label in traditional_model.classes_:
    print("-", label)

print("\nBoth Models Loaded Successfully!")
```

```python

# ============================================================
# Step 7.2 - Vehicle Component Prediction Function
# ============================================================

import pandas as pd

def predict_component(description):

    # 1. Validate input
    if not isinstance(description, str) or not description.strip():
        print("Please enter a valid defect description.")
        return None

    # 2. Clean whitespace (same approach as training)
    text = " ".join(description.split())

    # 3. Traditional model prediction
    traditional_pred = traditional_model.predict(
        [text]
    )[0]

    # 4. Generate MiniLM embedding
    embedding = minilm_encoder.encode(
        [text],
        convert_to_numpy=True,
        normalize_embeddings=minilm_config[
            "normalize_embeddings"
        ],
        show_progress_bar=False
    )

    # 5. MiniLM classifier prediction
    minilm_pred = minilm_classifier.predict(
        embedding
    )[0]

    # 6. Organize results
    results = pd.DataFrame({
        "Model": [
            "TF-IDF + Logistic Regression",
            "MiniLM + Logistic Regression"
        ],
        "Predicted Component": [
            traditional_pred,
            minilm_pred
        ]
    })

    # 7. Display results
    print("=" * 55)
    print("VEHICLE RECALL COMPONENT PREDICTION")
    print("=" * 55)

    print("\nInput Description:")
    print(text)

    print("\nPrediction Results:")
    display(results)

    return results


print("Prediction function created successfully!")
```

```python

# ============================================================
# Step 7.2 - Test Prediction Function
# ============================================================

# Test 1: Steering
print("\nTEST 1 - STEERING")

predict_component(
    "The driver may experience difficulty "
    "turning the wheel while driving."
)

# Test 2: Service Brakes
print("\nTEST 2 - BRAKES")

predict_component(
    "The vehicle may require a longer "
    "distance to stop."
)

# Test 3: Fuel System
print("\nTEST 3 - FUEL SYSTEM")

predict_component(
    "A damaged fuel line may cause "
    "gasoline to leak."
)
```

```python

# ============================================================
# Step 7.2 - Empty Input Validation
# ============================================================

result = predict_component("   ")

assert result is None

print("\nEmpty Input Handling: PASSED!")
```

```python

# ============================================================
# Step 7.3 - Interactive Vehicle Recall Classifier
# ============================================================

import ipywidgets as widgets
from IPython.display import display, HTML

# Title
title = widgets.HTML(
    value="""
    <h2>Vehicle Recall NLP Classifier</h2>
    <p>Enter a vehicle defect description to compare
    the predictions of both NLP models.</p>
    """
)

# Text input
description_input = widgets.Textarea(
    value="",
    placeholder="Enter a vehicle defect description...",
    description="Description:",
    layout=widgets.Layout(
        width="100%",
        height="150px"
    ),
    style={"description_width": "initial"}
)

# Prediction button
predict_button = widgets.Button(
    description="Predict Component",
    button_style="primary",
    icon="search",
    layout=widgets.Layout(width="220px")
)

# Output area
output = widgets.Output()

# Prediction event
def on_predict_clicked(button):

    predict_button.disabled = True

    try:
        output.clear_output(wait=True)

        with output:

            text = description_input.value

            if not text.strip():
                print("Please enter a valid defect description.")
                return

            # Use saved prediction function
            results = predict_component(text)

            if results is None:
                return

            # Compare model predictions
            predictions = results[
                "Predicted Component"
            ].tolist()

            print("\n" + "=" * 50)

            if predictions[0] == predictions[1]:
                print("MODEL AGREEMENT: YES")
                print("Both models predicted:", predictions[0])
            else:
                print("MODEL AGREEMENT: NO")
                print("The two models predicted different components.")

            print(
                "\nNote: This system classifies recall descriptions; "
                "it is not a vehicle diagnostic tool."
            )

    finally:
        predict_button.disabled = False


# Connect button to function
predict_button.on_click(on_predict_clicked)

# Display interface
interface = widgets.VBox([
    title,
    description_input,
    predict_button,
    output
])

display(interface)

print("Interactive Interface: READY!")
```
